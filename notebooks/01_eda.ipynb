# EDA — synthetic sensor log
Run `python -m data.synthetic_generator` and `python -m src.data_pipeline` first, then explore `data/processed/features.csv`.

```python
import pandas as pd
df = pd.read_csv("../data/processed/features.csv", parse_dates=["timestamp"])
df.info()
df.head()
```

```python
df.groupby("plot_id").agg(rows=("timestamp", "count"),
                           irrigation_rate=("irrigated_next_flag", "mean"),
                           sm_mean=("soil_moisture_pct", "mean"),
                           water_mean=("water_applied_l", "mean")).round(3)
```

```python
df.groupby("crop_growth_stage").agg(rows=("timestamp", "count"),
                                      sm_min=("soil_moisture_pct", "min"),
                                      sm_mean=("soil_moisture_pct", "mean")).round(1)
```

```python
pivot = df.pivot_table(index="timestamp", columns="plot_id", values="soil_moisture_pct")
pivot.plot(figsize=(14, 4), title="Soil moisture over time")
```
